# 01 - Cohort Exploration & Missingness Analysis

This notebook explores the clinical cohort from the UCI Chronic Kidney Disease repository (#336).


```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Add src to pythonpath
sys.path.insert(0, str(Path('../src')))
from diabetes_prediction.data.load_data import load_raw_dataset

df_raw = load_raw_dataset(uci_id=336, local_cache_path='../data/raw/chronic_kidney_disease.csv')
print(f'Raw dataset shape: {df_raw.shape}')
df_raw.head()
```

## Missing Value Profiling


```python
missing_counts = df_raw.isnull().sum().sort_values(ascending=False)
missing_pct = (missing_counts / len(df_raw) * 100).round(1)
pd.DataFrame({'Missing Count': missing_counts, 'Percent (%)': missing_pct}).head(12)
```

## Target Prevalence


```python
print('Diabetes target distribution:')
print(df_raw['dm'].value_counts(dropna=False))
print('\nHypertension target distribution:')
print(df_raw['htn'].value_counts(dropna=False))
```
